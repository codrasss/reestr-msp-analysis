```python
import pandas as pd 
df = pd.read_csv('filtered_construction.csv')
print('Размер:', df.shape)
print()
print('Типы данных:')
print(df.dtypes)
```

```python
df
```

```python
print('Пропущенных ИНН:', df['ИНН'].isna().sum())
print()
print('Дублей по ИНН:', df['ИНН'].duplicated().sum())
print()
inn_lengths = df['ИНН'].astype(str).str.len()
print('\nРаспределение длины ИНН:')
print(inn_lengths.value_counts())
print()
print("\nПропусков в 'Дата исключения из реестра':", df["Дата исключения из реестра"].isna().sum(), 'из', len(df))
```

```python
df['ИНН'] = df['ИНН'].astype(str).str.zfill(10)
print(df['ИНН'].str.len().value_counts())
print()
print(df['ИНН'].head())
```

```python
df['Дата включения в реестр'] = pd.to_datetime(df['Дата включения в реестр'], format ='%d.%m.%Y')

print('Самая ранняя дата:',df['Дата включения в реестр'].min())
print('Самая поздняя дата:', df['Дата включения в реестр'].max())

snapshot_date = pd.Timestamp('2024-04-15')
df['лет в реестре'] = (snapshot_date - df['Дата включения в реестр']).dt.days/365.25
print('\nСтатистика по возрасту( в годах ):')
print(df['лет в реестре'].describe())






                                               
```

```python
top_dates = df["Дата включения в реестр"].value_counts().head(10)
print(top_dates)
```

```python
top_regions = df["Регион"].value_counts().head(15)
print(top_regions)
```

```python
sample_df = df.sample(n=5000, random_state=42)

print("Размер выборки:", sample_df.shape)

sample_df.to_csv("sample_for_parsing.csv", index=False)
print("Сохранено в sample_for_parsing.csv")
```
